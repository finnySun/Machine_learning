## 一、生活例子：為什麼要畫圖？

假設你是咖啡店店長，這是這週每天賣出的杯數：

| 星期 | 一 | 二 | 三 | 四 | 五 | 六 | 日 |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- | :--- |
| 杯數 | 50 | 60 | 45 | 80 | 100 | 120 | 90 |

請你在 **3 秒內** 回答：「這週的生意是越來越好，還是越來越差？」

看數字，你需要一個一個比較。
但如果畫成一條線，**一眼就看得出來**：週三最低，週六最高，週末明顯比平日好。

**人類的眼睛很擅長看「形狀」，但不擅長看一堆數字。** 這就是資料視覺化（Data Visualization）的目的：

> 把數字變成圖，讓人一眼看出趨勢、差異和關係。

### 機器學習為什麼需要畫圖？
在訓練模型之前，資料分析師一定會先畫圖看資料，因為：

| 用途 | 例子 | 連結到哪一堂 |
| :--- | :--- | :--- |
| 看趨勢 | 營業額是不是逐月上升？ | 第 9 堂 預測數值 |
| 看關係 | 坪數越大，房價是不是越高？ | 第 9 堂 線性迴歸 |
| 抓異常值 | 有沒有一個點離大家特別遠？ | 第 6 堂 異常值 |
| 檢查清理結果 | 清理完的資料看起來合理嗎？ | 第 6 堂 資料清理 |

老師提醒：**先看圖，再建模型。** 如果連資料長什麼樣子都不知道，模型預測錯了你也看不出來。

## 二、概念解說

### 1. Matplotlib 是什麼？
- **Matplotlib**：Python 最常用的畫圖工具箱。
- **pyplot**：Matplotlib 裡面專門用來「畫圖」的工具，用法很像在畫布上一筆一筆畫。
- 慣用簡稱是 **plt**，就像 NumPy 叫 `np`、Pandas 叫 `pd`。

```python
import matplotlib.pyplot as plt
```

| 程式碼 | 意思 |
| :--- | :--- |
| `import` | 匯入 |
| `matplotlib.pyplot` | Matplotlib 工具箱裡的 pyplot 畫圖工具 |
| `as plt` | 取簡稱叫 plt |

到目前為止的三個簡稱，一起記：

| 工具箱 | 簡稱 | 用途 |
| :--- | :--- | :--- |
| NumPy | `np` | 處理數字陣列 |
| Pandas | `pd` | 處理表格 |
| Matplotlib.pyplot | `plt` | 畫圖 |

### 2. 三種圖怎麼選？（今天最重要的觀念）

| 圖的種類 | 英文 | 適合回答的問題 | 例子 | 指令 |
| :--- | :--- | :--- | :--- | :--- |
| 折線圖 | Line Chart | 隨著**時間**怎麼變化？ | 每月營業額、每天氣溫 | `plt.plot()` |
| 長條圖 | Bar Chart | 不同**類別**誰多誰少？ | 各商品銷量、各班平均分數 | `plt.bar()` |
| 散佈圖 | Scatter Plot | 兩個**數值**有沒有關係？ | 坪數 vs 房價、廣告費 vs 銷售額 | `plt.scatter()` |

記憶口訣：
- **折線看趨勢**（時間 → 線連起來）
- **長條比大小**（類別 → 一根一根比）
- **散佈看關係**（兩個數字 → 一點一點看）

### 3. 一張圖的組成

| 名稱 | 英文 | 說明 | 指令 |
| :--- | :--- | :--- | :--- |
| 標題 | Title | 這張圖在講什麼 | `plt.title()` |
| X 軸 | X-axis | 橫的那條軸 | `plt.xlabel()` |
| Y 軸 | Y-axis | 直的那條軸 | `plt.ylabel()` |
| 圖例 | Legend | 有多條線時，說明哪條是哪條 | `plt.legend()` |
| 顯示 | Show | 把圖畫出來 | `plt.show()` |

小技巧：**X 軸放「原因／時間／類別」，Y 軸放「結果／數量」。** 例如 X 是坪數，Y 是房價。

### 步驟 1：第一張折線圖

把剛剛的咖啡店一週銷量畫出來。

````python
import matplotlib.pyplot as plt

days = ["一", "二", "三", "四", "五", "六", "日"]
cups = [50, 60, 45, 80, 100, 120, 90]

plt.plot(days, cups)
plt.show()
````

#### 逐行解析

| 程式碼 | 意思 |
| :--- | :--- |
| `import matplotlib.pyplot as plt` | 匯入畫圖工具，簡稱 plt |
| `days = [...]` | X 軸的資料：星期幾 |
| `cups = [...]` | Y 軸的資料：杯數 |
| `plt.plot(days, cups)` | 畫折線圖，**第一個放 X、第二個放 Y** |
| `plt.show()` | 把圖顯示出來 |

#### 重要觀念：X 和 Y 的數量必須一樣多
`days` 有 7 個、`cups` 也有 7 個，電腦才能一對一配對：（一, 50）、（二, 60）……

這和第 3 堂的 shape 是同一件事：兩個 shape 都是 `(7,)` 才配得起來。
如果數量不同，會出現錯誤：`ValueError: x and y must have same first dimension`。

### 步驟 2：加上標題和軸名稱

上面那張圖沒有說明，別人看不懂 X 軸、Y 軸是什麼。一張好的圖一定要有標題和軸名稱。

````python
plt.plot(days, cups, marker="o", color="brown")
plt.title("咖啡店一週銷量")
plt.xlabel("星期")
plt.ylabel("杯數")
plt.show()
````

#### 逐行解析

| 程式碼 | 意思 |
| :--- | :--- |
| `marker="o"` | 在每個資料點畫一個圓點，方便看出實際的值 |
| `color="brown"` | 線的顏色設為咖啡色 |
| `plt.title("...")` | 圖的標題 |
| `plt.xlabel("...")` | X 軸名稱 |
| `plt.ylabel("...")` | Y 軸名稱 |

常用的 marker 和 color：

| marker | 樣子 | color | 顏色 |
| :--- | :--- | :--- | :--- |
| `"o"` | 圓點 ● | `"red"` | 紅 |
| `"s"` | 方塊 ■ | `"blue"` | 藍 |
| `"^"` | 三角形 ▲ | `"green"` | 綠 |
| `"x"` | 叉叉 ✕ | `"gray"` | 灰 |

老師提醒：`plt.title()`、`plt.xlabel()` 這些都要寫在 `plt.show()` **之前**。`plt.show()` 就像按下「列印」，按完之後再加的東西不會出現在這張圖上。

### 步驟 3：兩條線放在同一張圖比較

如果想比較「美式」和「拿鐵」兩種咖啡的銷量，可以畫兩條線，再用 **圖例（legend）** 說明哪條是哪條。

````python
americano = [30, 35, 25, 45, 60, 70, 50]
latte     = [20, 25, 20, 35, 40, 50, 40]

plt.plot(days, americano, marker="o", label="美式")
plt.plot(days, latte, marker="s", label="拿鐵")
plt.title("美式 vs 拿鐵 一週銷量")
plt.xlabel("星期")
plt.ylabel("杯數")
plt.legend()
plt.show()
````

#### 逐行解析

| 程式碼 | 意思 |
| :--- | :--- |
| 兩行 `plt.plot(...)` | 在同一張圖上畫兩條線 |
| `label="美式"` | 幫這條線取名字，給圖例使用 |
| `plt.legend()` | 顯示圖例，依照每條線的 label 自動產生 |

注意：只寫 `label` 但沒寫 `plt.legend()`，圖例不會出現；只寫 `plt.legend()` 但沒寫 `label`，圖例會是空的。**兩個要一起用。**

從圖可以看出：兩種咖啡的趨勢一樣（都是週六最高），但美式每天都賣得比拿鐵多。

### 步驟 4：長條圖 —— 比較不同類別

延續第 6 堂清理好的咖啡店資料，比較各商品的總銷量。商品是「類別」，不是時間，所以用長條圖。

````python
products = ["咖啡", "奶茶", "紅茶", "拿鐵"]
sales = [100, 150, 50, 80]

plt.bar(products, sales, color="orange")
plt.title("各商品銷量")
plt.xlabel("商品")
plt.ylabel("銷量")
plt.show()
````

#### 逐行解析

| 程式碼 | 意思 |
| :--- | :--- |
| `plt.bar(products, sales)` | 畫長條圖，第一個放類別（X），第二個放數值（Y） |
| `color="orange"` | 長條顏色 |

**為什麼這裡不用折線圖？**

折線圖會把點連起來，暗示「咖啡 → 奶茶 → 紅茶」之間有連續的變化。但商品之間沒有先後順序，連線是沒有意義的。

> 判斷方法：X 軸的東西「調換順序」後意思會不會變？
> - 星期一、二、三調換順序 → 意思變了 → **折線圖**
> - 咖啡、奶茶、紅茶調換順序 → 意思不變 → **長條圖**

### 步驟 5：散佈圖 —— 看兩個數值的關係

回到第 1、2 堂的房價例子。我們有 8 間房子的坪數和房價，想知道：**坪數和房價有沒有關係？**

````python
size  = [20, 25, 30, 35, 40, 45, 50, 32]
price = [500, 640, 760, 860, 1010, 1120, 1250, 2000]

plt.scatter(size, price, color="blue")
plt.title("坪數與房價的關係")
plt.xlabel("坪數")
plt.ylabel("房價（萬）")
plt.show()
````

#### 逐行解析

| 程式碼 | 意思 |
| :--- | :--- |
| `plt.scatter(size, price)` | 畫散佈圖，每一間房子是一個點，位置是（坪數, 房價） |

#### 從這張圖你可以看到兩件事

**1. 大部分的點排成一條往右上的直線**
→ 坪數越大，房價越高，兩者有「正相關」。
→ 第 9 堂的**線性迴歸**，就是讓電腦找出這條直線，再用它預測新房子的價格（第 2 堂的 `fit` 和 `predict`）。

**2. 有一個點特別高（32 坪、2000 萬）**
→ 它離其他點很遠，這就是第 6 堂的**異常值**。
→ 可能是輸入錯誤，也可能是特殊物件（例如市中心豪宅）。
→ 用數字不容易發現，但畫成圖**一眼就看到了**。

這就是「先看圖，再建模型」的意思：如果沒先畫圖就直接訓練模型，這個異常值會把模型學到的直線拉歪。

### 步驟 6：搭配 Pandas —— 直接把表格畫成圖

實務上，資料通常是用 `pd.read_csv()` 讀進來的 DataFrame。
畫圖時只要**選出欄位**（第 5 堂學的 `df["欄位名稱"]`），放進 `plt.plot()` 就可以了。

這裡先用字典建立資料，實務上換成 `pd.read_csv("檔名.csv")` 即可。

````python
import pandas as pd

df = pd.DataFrame({
    "月份": ["1月", "2月", "3月", "4月", "5月", "6月"],
    "營業額": [120, 135, 128, 160, 175, 190]
})

print(df)

plt.plot(df["月份"], df["營業額"], marker="o", color="green")
plt.title("上半年營業額")
plt.xlabel("月份")
plt.ylabel("營業額（萬）")
plt.show()
````

#### 逐行解析

| 程式碼 | 意思 |
| :--- | :--- |
| `pd.DataFrame({...})` | 建立表格（第 5 堂） |
| `df["月份"]` | 取出月份這一欄，是一個 Series，當作 X |
| `df["營業額"]` | 取出營業額這一欄，當作 Y |
| `plt.plot(X, Y)` | 和前面一樣畫折線圖 |

**重點：** 放進 `plt.plot()` 的可以是 Python List、NumPy 陣列，也可以是 Pandas 的 Series，用法完全一樣。

### 步驟 7：結合第 6 堂 —— 清理完再畫圖

資料有缺失值時，畫出來的線會斷掉。所以流程是：**清理 → 再畫圖**。

````python
import numpy as np

df2 = pd.DataFrame({
    "月份": ["1月", "2月", "3月", "4月", "5月", "6月"],
    "營業額": [120, 135, np.nan, 160, 175, 190]
})

# 清理前：3 月是 NaN
plt.plot(df2["月份"], df2["營業額"], marker="o", label="清理前")

# 清理：用平均值填補（第 6 堂）
df2["營業額"] = df2["營業額"].fillna(df2["營業額"].mean())

# 清理後
plt.plot(df2["月份"], df2["營業額"], marker="x", linestyle="--", label="清理後")

plt.title("清理前後比較")
plt.xlabel("月份")
plt.ylabel("營業額（萬）")
plt.legend()
plt.show()

print("3 月填補的值：", df2["營業額"][2])
````

#### 逐行解析

| 程式碼 | 意思 |
| :--- | :--- |
| 第一次 `plt.plot` | 畫清理前的資料，NaN 的位置畫不出來，線會斷掉 |
| `fillna(df2["營業額"].mean())` | 用平均值填補缺失值（第 6 堂） |
| `linestyle="--"` | 線改成虛線，方便和清理前區分 |
| `df2["營業額"][2]` | 取出索引 2（也就是 3 月）的值 |

觀察：清理前的線在 3 月斷掉；清理後補上了 156 萬（其他五個月的平均）。
畫圖可以幫你**檢查清理的結果合不合理**：如果補上的值讓線突然暴衝，就要回頭檢查填補的方法。

### 練習四（挑戰題）：清理 + 畫圖
下面的資料有兩個問題：價格欄有一筆是文字「錯誤」，銷量欄有缺失值。
請完成：
1. 用 `pd.to_numeric(..., errors="coerce")` 把價格轉成數字。
2. 用銷量的平均值填補缺失值。
3. 畫出各商品銷量的**長條圖**。

````python
# 練習四：資料（先執行這格）
shop = pd.DataFrame({
    "商品": ["咖啡", "奶茶", "紅茶", "拿鐵"],
    "價格": ["80", "60", "40", "錯誤"],
    "銷量": [100, np.nan, 50, 80]
})
print(shop)

plt.bar(shop["商品"], shop["銷量"], color = "skyblue")
plt.xlabel("商品")
plt.ylabel("銷量")
plt.title("商品銷售關係圖")
plt.show
````

````python
# 練習四：在這裡寫你的程式
shop["價格"] = pd.to_numeric(shop["價格"], errors="coerce")
shop["銷量"] = shop["銷量"].fillna(shop["銷量"].mean())
print(shop)

plt.bar(shop["商品"], shop["銷量"], color = "skyblue")
plt.xlabel("商品")
plt.ylabel("銷量")
plt.title("商品銷售關係圖")
plt.show
````

````python
# 練習四 參考解答
shop["價格"] = pd.to_numeric(shop["價格"], errors="coerce")
shop["銷量"] = shop["銷量"].fillna(shop["銷量"].mean())
print(shop)

plt.bar(shop["商品"], shop["銷量"], color="teal")
plt.title("各商品銷量（清理後）")
plt.xlabel("商品")
plt.ylabel("銷量")
plt.show()
````
