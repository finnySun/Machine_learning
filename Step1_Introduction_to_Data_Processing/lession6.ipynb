### 第二部分：缺失值（Missing Values）

#### 1. 什麼是缺失值？
英文：Missing Values

中文：缺失值、遺漏值。

簡單說，就是某個欄位沒有資料。

例如：
| 姓名 | 年齡 | 消費金額 |
| :--- | ---: | ---: |
| 小明 | 25 | 1000 |
| 小美 | 30 | 2000 |
| 小華 | NaN | 1500 |

這裡的 NaN 就代表沒有有效的數值。

NaN 是 Not a Number 的縮寫，在 Pandas 中常用來表示缺失資料。

#### 2. 如何找出缺失值？
先建立資料：

````python
import pandas as pd
import numpy as np # 使用 np.nan 建立缺失值。

df = pd.DataFrame({
    "姓名": ["小明", "小美", "小華", "小林"],
    "年齡": [25, 30, np.nan, 35],
    "消費金額": [1000, 2000, 1500, np.nan]
})

print(df)
````

#### 3. isnull()：找出缺失值

````python
print(df.isnull())
````

怎麼看？
| 結果 | 意思 |
| :--- | :--- |
| True | 這個位置有缺失值 |
| False | 這個位置沒有缺失值 |

**isnull() 就是檢查資料有沒有空缺**

#### 4. sum()：計算缺失值數量
剛剛我們學過 sum()，現在把它用在 Pandas！

````python
print(df.isnull().sum())
````

代表：
- 姓名：0 個缺失值。
- 年齡：1 個缺失值。
- 消費金額：1 個缺失值。

這是實務上非常常用的寫法。

#### 5. fillna()：填補缺失值

英文：Fill Missing Values

中文：填補缺失值。

例如，小華沒有年齡資料。

我們可以用平均年齡填補。

````python
df["年齡"] = df["年齡"].fillna(df["年齡"].mean())
````

逐行解釋：

| 程式碼 | 意思 |
| :--- | :--- |
| `df["年齡"]` | 選取年齡欄位 |
| `.mean()` | 計算年齡平均值 |
| `.fillna()` | 把缺失值填補起來 |
| `=` | 把處理後的結果存回年齡欄位 |

原本年齡：
25, 30, NaN, 35

平均值：
(25+30+35)/ 3=30

填補後：
25, 30, 30, 35

另外還可以使用固定數字：
```df["年齡"] = df["年齡"].fillna(0)```

或使用中位數：
```df["年齡"] = df["年齡"].fillna(df["年齡"].median())```

#### 6. dropna()：刪除缺失值

英文：Drop Missing Values

中文：刪除缺失值。

如果某筆資料缺失太多，或這筆資料無法使用，也可以考慮刪除。

```df = df.dropna()```

意思是：

把含有缺失值的資料列刪除。

注意：`dropna()` 預設會刪除至少有一個缺失值的資料列。

老師提醒： 不是看到缺失值就全部刪除。要先判斷資料的重要性，以及缺失原因。

### 第三部分：重複資料（Duplicate Data）
#### 1. 什麼是重複資料？
英文：Duplicate Data

中文：重複資料。

例如：


| 姓名 | 年齡 | 消費金額 |
| :--- | ---: | ---: |
| 小明 | 25 | 1000 |
| 小美 | 30 | 2000 |
| 小明 | 25 | 1000 |
| 小華 | 28 | 1500 |

小明出現兩次，而且資料完全相同。

如果我們計算客戶數量，就可能把小明算成兩個人。

#### 2. duplicated()：找出重複資料

```print(df.duplicated())```

結果可能是：
- 0    False
- 1    False
- 2     True
- 3    False

代表：
- False：不是重複資料。
- True：這一列是重複資料。

Pandas 預設會保留第一次出現的資料，後續相同的資料會標記為重複。

#### 3. drop_duplicates()：刪除重複資料

```df = df.drop_duplicates()```


### 第四部分：異常值（Outliers）
### 1. 什麼是異常值？
英文：Outliers

中文：異常值、離群值。

就是和大部分資料差異很大的數值。

例如：

| 客戶 | 消費金額 |
| :--- | ---: |
| 小明 | 1000 |
| 小美 | 1200 |
| 小華 | 1500 |
| 小林 | 1300 |
| 小王 | 999999 |

小王的消費金額明顯比其他人高很多。

這可能是：
- 輸入錯誤。
- 特殊大額交易。
- 真正的高消費客戶。
所以不能看到數字很大，就直接刪除。

#### 2. 如何初步觀察異常值？
我們可以使用：

```print(df.describe())```

觀察：
- min：最小值。
- max：最大值。
- mean：平均值。
- 50%：中位數。

如果最大值與其他資料差距非常大，就值得進一步調查。

#### 3. 使用 IQR 找出異常值

英文：Interquartile Range

中文：四分位距。

這是常見的異常值偵測方法。

先認識三個數字：
- Q1：第 25 百分位數。
- Q2：第 50 百分位數，也就是中位數。
- Q3：第 75 百分位數。

四分位距：
**IQR=Q3-Q1**

常見的異常值判斷範圍：
- **下界=Q1-1.5 * IQR**
- **上界=Q3+1.5 * IQR**

超過上下界的資料，會被標記為潛在異常值。

先不用急著背公式，我們用程式操作：


````python
Q1 = df["消費金額"].quantile(0.25)
Q3 = df["消費金額"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["消費金額"] < lower) |
    (df["消費金額"] > upper)
]

print(outliers)
````

這裡第一次看到一個新符號：

| 代表 OR（或），只要符合左邊或右邊其中一個條件，就會被選出來。

注意：IQR 是一種常見的偵測方法，不代表被標記的資料一定錯誤。還是要回到實際情境判斷。

### 第五部分：資料型態錯誤（Data Type Errors）
這部分和上一堂的 dtype 有關。

假設你有一份價格資料：


| 商品 | 價格 |
| :--- | ---: |
| 咖啡 | 80 |
| 奶茶 | 60 |
| 紅茶 | 40 |

正常情況下，價格應該是數字。

但有時候資料匯入後，可能變成文字。

例如：

```price = pd.Series(["80", "60", "40"])```

```print(price.dtype)```

結果：object

object 是 Pandas 常見的資料型態標示，可能代表文字或混合型態。

如果要轉成數字，可以使用：

```price = pd.to_numeric(price)```

轉換後：

- 0    80
- 1    60
- 2    40
- dtype: int64

**遇到無法轉換的資料怎麼辦？**

例如：```price = pd.Series(["80", "60", "錯誤", "40"])```

直接轉換可能會出錯。

這時候可以使用：

```price = pd.to_numeric(price, errors="coerce")```

errors="coerce" 的意思是：

遇到無法轉換的資料，就把它轉成 NaN。

結果：
- 0    80.0
- 1    60.0
- 2     NaN
- 3    40.0

之後就可以再使用 `fillna()` 或其他方式處理。

### 第六部分：清理後儲存資料
前面清理完資料後，通常需要將結果儲存。

使用：
```df.to_csv("cleaned_data.csv", index=False)```

逐行解釋：

| 程式碼 | 意思 |
| :--- | :--- |
| `df` | 已經整理好的資料 |
| `to_csv()` | 儲存成 CSV |
| `"cleaned_data.csv"` | 儲存檔案名稱 |
| `index=False` | 不把 Pandas 的索引另外存成一欄 |

這樣就能把整理後的資料輸出成新的 CSV 檔案。


### 第七部分：完整實作練習

現在我們把今天學到的技能串起來。

假設你是咖啡店的資料分析師，主管給你一份銷售資料。

我們要完成：
1. 建立資料。
2. 檢查缺失值。
3. 填補缺失值。
4. 刪除重複資料。
5. 修正資料型態。
6. 查看整理後的資料。

````python
# 第一步：建立原始資料
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "商品": ["咖啡", "奶茶", "紅茶", "咖啡", "拿鐵"],
    "價格": ["80", "60", "40", "80", "錯誤"],
    "銷量": [100, 150, np.nan, 100, 80]
})

print(df)
````

````python
# 第二步：檢查缺失值

print(df.isnull().sum())
````

````python
# 第三步：修正價格資料型態
df["價格"] = pd.to_numeric(
    df["價格"],
    errors="coerce"
)

# 原本的 "錯誤" 就會轉成 NaN。
````

````python
# 第四步：填補缺失值

df["銷量"] = df["銷量"].fillna(
    df["銷量"].mean()
)
````

````python
# 第五步：刪除重複資料

df = df.drop_duplicates()
````

````python
# 第六步：檢查整理結果

print(df)
print(df.info())
````

這就是一個簡單的資料清理流程。

### 資料清理流程
**原始資料**
收到 CSV 或 Excel 資料

↓

**檢查缺失值**
使用 `isnull()`

↓

**填補或刪除缺失值**
使用 `fillna()`、`dropna()`

↓

**檢查重複資料**
使用 `duplicated()`、`drop_duplicates()`

↓

**檢查異常值與型態**
使用 `describe()`、`to_numeric()`

↓

**輸出整理後資料**
使用 `to_csv()`
