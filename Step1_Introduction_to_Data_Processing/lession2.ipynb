#### 二、第一種：監督式學習（Supervised Learning）
Supervised Learning

##### 1. 什麼是監督式學習？
先拆解英文：

- Supervised：有人監督、有人指導。
- Learning：學習。

合起來就是：有人提供正確答案，讓電腦學習。

想像你是一位老師，正在教學生認識水果。

你拿出水果照片：
| 圖片 | 告訴電腦的答案 |
| :--- | :--- |
| 蘋果照片 | 蘋果 |
| 香蕉照片 | 香蕉 |
| 柳橙照片 | 柳橙 |

電腦一邊看照片，一邊知道正確答案。

經過大量練習後，你拿出一張新的香蕉照片，電腦就能判斷：

「這應該是香蕉！」

這就是監督式學習。

####　2. 監督式學習有兩種常見任務
這裡很重要，請特別注意。

##### A. 分類（Classification）

分類就是：把資料分到不同類別。

例如：
- 這封 Email 是垃圾郵件還是正常郵件？
- 這張照片是貓還是狗？
- 這筆交易是正常還是詐騙？

白話：答案是某一種類別。

##### B. 迴歸（Regression）

迴歸就是：預測一個數字。

例如：
- 預測房價。
- 預測明天的氣溫。
- 預測下個月的銷售額。

白話：答案是一個數值。

| 比較 | 分類 | 迴歸 |
| :--- | :--- | :--- |
| 英文 | Classification | Regression |
| 預測結果 | 類別 | 數字 |
| 例子 | 貓或狗 | 房價 1,500 萬 |
| 例子 | 垃圾或正常郵件 | 銷售額 50 萬 |


記住：

分類 = 選類別。

迴歸 = 預測數字。

#### 三、第二種：非監督式學習（Unsupervised Learning）
Unsupervised Learning
##### 1. 什麼是非監督式學習？

拆解英文：
- Unsupervised：沒有提供答案監督。
- Learning：學習。
意思就是：

我們不提供正確答案，讓電腦自己從資料中尋找規律。

假設你經營一家咖啡店，有 1,000 位顧客。

你想知道這些顧客有哪些不同的消費習慣。

但是，你不知道他們屬於哪一種類型。

你只提供電腦以下資料：

| 顧客 | 每月消費次數 | 平均消費金額 |
| :--- | :--- | :--- |
| A | 20 次 | 150 元 |
| B | 18 次 | 160 元 |
| C | 2 次 | 80 元 |
| D | 3 次 | 90 元 |
| E | 15 次 | 500 元 |


電腦可能發現：

- A、B 消費頻率高，金額中等。
- C、D 消費頻率低，金額較低。
- E 消費頻率高，金額也高。

於是電腦把相似的顧客分成不同群組。

這種方法叫做：

分群（Clustering）

##### 2. 生活中有哪些應用？
- 電商平台：找出消費習慣相似的顧客。
- 超商：分析不同消費族群。
- 銀行：分析客戶交易行為。
- 行銷：將顧客分成不同客群，設計行銷策略。
例如，你在蝦皮購物，平台可能根據你的購物行為，將你歸類到某個消費族群。

注意：非監督式學習本身不會事先知道群組的正確名稱，群組代表什麼意義，通常還需要人類進一步分析。

#### 四、第三種：強化學習（Reinforcement Learning）
Reinforcement Learning
##### 1. 什麼是強化學習？
先想像一個情境：

你正在教一隻狗狗學習握手。

過程是這樣：
1. 你叫狗狗握手。
2. 狗狗嘗試抬起前腳。
3. 如果成功，你給牠零食獎勵。
4. 如果沒有成功，就不給獎勵。
5. 狗狗經過多次嘗試，慢慢知道怎麼做才能得到獎勵。
這就是強化學習的概念。

電腦也可以透過類似方式學習。
##### 2. 強化學習的四個重要名詞

| 英文 | 中文 | 白話意思 |
| :--- | :--- | :--- |
| Agent | 代理人 | 負責學習、做決策的角色 |
| Environment | 環境 | Agent 所處的世界 |
| Action | 行動 | Agent 做出的選擇 |
| Reward | 獎勵 | 做出選擇後得到的回饋 |


例如：訓練 AI 玩遊戲。
- Agent：玩遊戲的 AI。
- Environment：遊戲地圖。
- Action：向左、向右、向前。
- Reward：抵達終點得到獎勵，撞到障礙物可能得到負分。
AI 透過反覆嘗試，逐漸找到比較好的策略。

##### 3. 強化學習的應用
- AI 玩遊戲。
- 機器人學習走路。
- 自動駕駛中的決策研究。
- 資源調度與策略最佳化。

五、三種學習方式一次比較
這張表建議你保存起來，之後考試很常需要分辨這些概念。

| 比較 | 監督式 | 非監督式 | 強化式 |

| 英文 | Supervised | Unsupervised | Reinforcement |

再教你一個簡單記憶法：

#### 六、第一次接觸機器學習實作
接下來我們開始看 Python。

今天使用一個很簡單的例子，讓你認識機器如何學習。

我們使用的是監督式學習中的「迴歸」。

範例：預測房價


```python

# 先安裝及匯入機器學習工具
from sklearn.linear_model import LinearRegression

# sklearn：Python 常用的機器學習工具箱。
# linear_model：線性模型的工具。
#　LinearRegression：線性迴歸模型。

# Step 1：準備資料
X = [[20], [30], [40], [50]]
y = [500, 750, 1000, 1250]

# X：輸入資料，也就是 Feature（特徵）。
# y：正確答案，也就是 Label（標籤）。


# Step 2：建立模型 (建立一個線性迴歸模型，並把它取名為 model)
model = LinearRegression()


# Step 3：訓練模型 (fit() 可以理解成「訓練模型」。
model.fit(X, y)

# 意思是：把坪數 X 和房價 y 提供給模型，讓模型學習兩者之間的關係。)


# Step 4：預測 60 坪房子的價格
prediction = model.predict([[60]])

# 意思是：我們提供一間 60 坪的新房子，請模型預測它的價格。

# Step 5：顯示結果
print(prediction)
```


| 程式碼 | 意思 |
| :--- | :--- |
| `LinearRegression()` | 建立模型 |
| `fit()` | 訓練模型 |
| `predict()` | 使用模型預測 |
