##### 一、ndim — 資料有幾個維度？
英文全名：Number of Dimensions

中文：維度數量。

```python
## 1. 一維陣列（1D Array）

import numpy as np

data = np.array([10, 20, 30, 40])

print(data.ndim)
```

```python
## 2. 二維陣列（2D Array）

data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(data.ndim)
```

##### 3. ndim 和 shape 有什麼不同？
這是今天第一個重要觀念！
|指令|功能|範例結果|
| :--- | :---: | ---: |
|shape|每個維度各有多少個元素|(2, 3)|
|ndim|總共有幾個維度|2|

### 第三部分：dtype — 資料的型態
英文全名：Data Type

中文：資料型態。

我們先看兩組資料：

```python
a = np.array([10, 20, 30])
b = np.array([10.5, 20.5, 30.5])

print(a.dtype)
print(b.dtype)
```

### 常見的資料型態
|型態|英文|意思|範例|
| :--- | :---: | ---: | ---: |
|整數|Integer|沒有小數點|10、20、30|
|浮點數|Float|可以有小數點|10.5、20.3|
|布林值|Boolean|是或否|True、False|
|字串|String|文字|"Apple"|

### 第四部分：sum() — 計算總和
這個很簡單，就是把數字全部加起來。

假設你有一份商店每日營業額：

```python
import numpy as np

sales = np.array([1000, 2000, 3000, 4000])

print(sales.sum())
```

### 第五部分：axis — 指定計算方向
英文：Axis

中文：軸、方向。

這個觀念一開始可能有點抽象，我們用成績表來理解。

假設有兩位學生，分別考了三科：
| 學生 | 國文 | 英文 | 數學 |
| :---: | :---: | :---: | :---: |
| A | 80 | 90 | 70 |
| B | 60 | 70 | 80 |

```python
score = np.array([
    [80, 90, 70],
    [60, 70, 80]
])

print(score.sum())
print(score.sum(axis=0))
print(score.sum(axis=1))
```

### 第六部分：mean() — 計算平均值
英文：Mean

中文：平均數。

例如：

```python
score = np.array([
    [80, 90, 70],
    [60, 70, 80]
])

print(score.mean())
print(score.mean(axis=0))
print(score.mean(axis=1))
```

### 第七部分：把今天的知識應用在機器學習
假設我們有一份房屋資料：
- 第一欄：房屋坪數
- 第二欄：房屋屋齡

```python
import numpy as np

X = np.array([
    [20, 5],
    [30, 10],
    [40, 15],
    [50, 20]
])

print(X.shape)
print(X.ndim)
print(X.dtype)
print(X.mean(axis=0))
```

| 程式碼 | 意思 |
| :--- | :--- |
| `X = np.array(...)` | 建立房屋資料 |
| `X.shape` | 有 4 筆資料、2 個欄位 |
| `X.ndim` | 這是二維資料 |
| `X.dtype` | 檢查資料型態 |
| `X.mean(axis=0)` | 計算坪數平均與屋齡平均 |

最後得到：
- 平均坪數：35 坪
- 平均屋齡：12.5 年
這就是機器學習很常見的資料處理方式。

### 第 4 堂課重點整理

| 指令 | 功能 | 記憶方式 |
| :--- | :--- | :--- |
| `shape` | 查看資料形狀 | 幾列幾欄 |
| `ndim` | 查看維度數量 | 有幾個維度 |
| `dtype` | 查看資料型態 | 整數還是小數 |
| `sum()` | 計算總和 | 全部加起來 |
| `mean()` | 計算平均 | 加總除以數量 |
| `axis=0` | 沿列方向計算 | 各欄位 |
| `axis=1` | 沿欄方向計算 | 各列 |
