#### 一、為什麼學機器學習要學 NumPy？

**NumPy 是什麼？**

**NumPy 是 Python 裡面一個非常常用的數值運算工具。**

英文全名是：

Numerical Python

你可以把它想成一個專門處理數字的工具箱。

它可以幫我們：

- 整理資料。

簡單來說：
NumPy 就像一台專門處理大量數字的計算機。

#### 二、先認識什麼是 Array（陣列）

NumPy 有自己的陣列，叫做 NumPy Array（NumPy 陣列）。

它跟 Python List 有什麼不同？
- List：可以存放不同種類的資料，例如數字、文字。
- NumPy Array：主要用來高效率地處理數值資料，特別適合大量數學運算。
我們可以把 List 想成一般收納盒，而 NumPy Array 想成專門設計給數字使用的高效率收納盒。

#### 三、第一個 NumPy 程式碼：np.array()

```python
import numpy as np
```

| 程式碼 | 意思 |
| :--- | :--- |
| `import` | 匯入、使用 |
| `numpy` | NumPy 工具箱 |
| `as` | 幫它取一個簡稱 |
| `np` | NumPy 的簡稱 |

所以整行意思是：

我要使用 `numpy`，並且把它簡稱為 `np`。

接下來：

```python
import numpy as np

price = np.array([100, 120, 150, 180, 200])

print(price)
```

這裡的：

```np.array()```

就是把資料轉換成  `numpy` 陣列。

執行結果：
[100 120 150 180 200]

NumPy 陣列顯示時，數字之間通常沒有逗號

#### 四、認識 Shape：資料的形狀
這是今天最重要的觀念之一。

shape 的意思是「形狀」。

我們先用生活例子理解。

假設你有一排 5 個蘋果：

```shape = (5,)```

如果我們有 2 排，每排 3 個蘋果：

```shape = (2, 3)```

所以：
- (5,)：一排 5 個。
- (2, 3)：2 排、3 欄。
- (3, 4)：3 排、4 欄。
shape 就是在告訴我們資料的排列方式。


##### 1. 怎麼知道陣列的 Shape？
我們可以使用 .shape。

````python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print(data.shape)
````

````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(data.shape)
````

為什麼是 (2, 3)？

因為：
- 第一個數字 2：有 2 排。
- 第二個數字 3：每排有 3 個數字。
老師提醒

shape 只是在描述資料的形狀，不是在計算資料的總和。

如果是 (2, 3)，代表 2 排、3 欄，總共有 6 個數字。

#### 五、認識 Reshape：改變陣列形狀
這個單字可能是你之後看機器學習程式碼最常遇到的其中一個。

reshape = 改變形狀。

例如：

原本我們有 6 個數字：

```data = np.array([1, 2, 3, 4, 5, 6])```

如果我想把它改成 2 排、每排 3 個，怎麼辦？

使用：

```data = data.reshape(2, 3)```

#### 六、為什麼機器學習需要 NumPy？
回到我們前面學過的房價預測。

假設有很多間房子的資料：


| 房子 | 坪數 | 屋齡 |
| :--- | :--- | :--- |
| A | 20 | 5 |
| B | 30 | 10 |
| C | 40 | 15 |
| D | 50 | 20 |

如果要讓電腦學習房價，就需要把資料整理成電腦能處理的數字形式。

老師帶你看：
- 4：有 4 間房子。
- 2：每間房子有 2 個特徵。

也就是：
- 第一個特徵：坪數。
- 第二個特徵：屋齡。
這裡就連結到第二堂課的 Feature（特徵）。

**機器學習會把資料整理成這種陣列，再交給模型進行學習。**

這也是為什麼我們需要 NumPy。



#### 七、補充：你可能會在程式碼中看到的三個符號


| 程式碼 | 意思 |

````python
import numpy as np

X = np.array([
    [20, 5],
    [30, 10],
    [40, 15],
    [50, 20]
])

print(X.shape)
````
