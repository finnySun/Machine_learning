#### 第三部分：認識 Series 和 DataFrame

這是今天最重要的兩個名詞。

##### 1. Series：一維資料

英文：Series

中文：序列。

你可以把它想成 Excel 裡面的一整欄資料。


````python
import pandas as pd

age = pd.Series([25, 30, 28, 35])

print(age)
````

左邊多了 0、1、2、3

這是 Pandas 自動建立的索引（Index），用來標示每筆資料的位置。


| 索引 Index | 資料 Value |
| :--- | :--- |
| 0 | 25 |
| 1 | 30 |
| 2 | 28 |
| 3 | 35 |


##### 2. DataFrame：二維資料

英文：DataFrame

中文：資料框。

你可以把它想成一張完整的 Excel 表格。

````python
import pandas as pd

data = {
    "姓名": ["小明", "小美", "小華"],
    "年齡": [25, 30, 28],
    "消費金額": [1000, 2000, 1500]
}

df = pd.DataFrame(data)

print(df)
````

我們逐行解釋：

##### 第一段：建立資料

```data = {```
    ```"姓名": ["小明", "小美", "小華"],```
    ```"年齡": [25, 30, 28],```
    ```"消費金額": [1000, 2000, 1500]```
```}```

這裡使用了 Python 的字典（Dictionary）。

字典可以想成一個有名稱的資料對照表：
- "姓名" 是欄位名稱。
- ["小明", "小美", "小華"] 是這個欄位的資料。
- "年齡" 是另一個欄位名稱。
- [25, 30, 28] 是對應的資料。

##### 第二段：建立 DataFrame

```df = pd.DataFrame(data)```

意思是：把剛剛的資料轉成 Pandas 的表格。

##### 第三段：顯示資料

```print(df)```

把表格印出來。

| Index | 姓名 | 年齡 | 消費金額 |
| :--- | :--- | :--- | :--- |
| 0 | 小明 | 25 | 1000 |
| 1 | 小美 | 30 | 2000 |
| 2 | 小華 | 28 | 1500 |

##### Series 和 DataFrame 的差別


| 比較 | Series | DataFrame |
| :--- | :--- | :--- |
| 維度 | 一維 | 二維 |
| 外觀 | 一欄資料 | 完整表格 |
| 類似 Excel | 單一欄 | 整張工作表 |
| 建立方式 | `pd.Series()` | `pd.DataFrame()` |

記憶方式：Series 是一欄，DataFrame 是一張表。


#### 第四部分：讀取 CSV 檔案
這個技能在資料分析和機器學習非常重要。

先認識 CSV：

英文：Comma-Separated Values

中文：逗號分隔值。

CSV 是一種常見的表格資料檔案格式，可以用 Excel 開啟，也可以讓 Python 讀取。

例如，
有一份檔案：customer.csv

資料:

name,age,amount

小明,25,1000

小美,30,2000

小華,28,1500

````python
import pandas as pd

df = pd.read_csv("customer.csv")

print(df)
````

逐行解釋：

| 程式碼 | 意思 |
| :--- | :--- |
| `import pandas as pd` | 匯入 Pandas |
| `pd.read_csv()` | 讀取 CSV 檔案 |
| `"customer.csv"` | 檔案名稱 |
| `df` | 儲存讀取後的表格 |

注意：這個 CSV 檔案必須放在 Python 程式可以找到的位置，否則會出現找不到檔案的錯誤。

如果你使用 Google Colab，通常可以先上傳檔案，再讀取它。


#### 第五部分：查看資料內容
假設我們已經讀取一份客戶資料：

````python
import pandas as pd

df = pd.DataFrame({
    "姓名": ["小明", "小美", "小華", "小林"],
    "年齡": [25, 30, 28, 35],
    "消費金額": [1000, 2000, 1500, 3000]
})
````

````python
# head()：查看預設顯示前 5 筆資料
print(df.head())
````

````python
# 想看前 2 筆
print(df.head(2))
````

````python
# tail()：查看最後幾筆資料(顯示最後 2 筆資料)
print(df.tail(2))
````

````python
# info()：查看資料資訊
df.info()
````

可以看到：
- 有多少筆資料。
- 有多少個欄位。
- 每個欄位的資料型態。
- 是否有缺失資料。

````python
# describe()：查看統計摘要
print(df.describe())
````

會幫你計算數值欄位的：
| 統計項目 | 意思 |
| :--- | :--- |
| `count` | 資料筆數 |
| `mean` | 平均值 |
| `std` | 標準差 |
| `min` | 最小值 |
| `25%` | 第 25 百分位數 |
| `50%` | 中位數 |
| `75%` | 第 75 百分位數 |
| `max` | 最大值 |


#### 第六部分：選取欄位
假設我們想知道所有客戶的年齡，不需要把整張表格都印出來。

只需要：

````python
print(df["年齡"])
````

````python
# 同時取得年齡和消費金額：
print(df[["年齡", "消費金額"]])
````

為什麼第二個有兩層中括號？
- 外層 []：告訴 Pandas 我要選欄位。
- 內層 []：用 Python 清單列出多個欄位名稱。

#### 第七部分：篩選資料
現在我們進入資料分析最常用的功能之一。

假設我們想找出年齡大於 28 歲的客戶。

````python
result = df[df["年齡"] > 28]

print(result)
````

逐行拆解：

```df["年齡"] > 28```

這一段是在問：

每一位客戶的年齡，有沒有大於 28 歲？

得到的結果會是：
- 0    False
- 1     True
- 2    False
- 3     True

True：符合條件。

False：不符合條件。

接著：

```df[df["年齡"] > 28]```

就是只保留符合條件的資料。
常見的篩選符號
| 符號 | 意思 | 範例 |
|---|---|---|
| > | 大於 | 年齡 > 28 |
| < | 小於 | 年齡 < 30 |
| >= | 大於等於 | 年齡 >= 30 |
| <= | 小於等於 | 年齡  數值 | 篩選資料 |
