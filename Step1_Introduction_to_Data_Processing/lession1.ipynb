#### 1. 先認識三個重要名詞
你可能常常聽到這三個詞：
- AI（人工智慧）
- Machine Learning（機器學習）
- Deep Learning（深度學習）
它們有什麼不同？

先看這張關係圖：
### AI 人工智慧
* **定義**：讓電腦執行看起來需要智慧的工作。

↓

### Machine Learning 機器學習
* **定義**：讓電腦從資料中學習規律。

↓

### Deep Learning 深度學習
* **定義**：利用多層神經網路學習複雜的規律。


這三者是包含關係：
- AI 是最大的範圍。
- 機器學習是 AI 的其中一種方法。
- 深度學習是機器學習的其中一種方法。

用生活例子理解
假設我們要讓電腦認識貓咪。


- AI： 我們希望電腦能夠辨識這是不是貓。

- 機器學習： 我們提供大量貓咪和狗狗的照片，讓電腦自己學習兩者的差異。

- 深度學習： 使用更複雜的神經網路，讓電腦從圖片中學習眼睛、耳朵、輪廓等特徵。

記住這句話：
機器學習就是讓電腦從資料中找規律，而不是每件事情都要人類寫好規則告訴它。

#### 2. 機器學習到底怎麼運作？

這是今天最重要的觀念。

假設我們想讓電腦預測房子的價格。

我們先給電腦一些資料：


| 房子 | 坪數 | 房價 |
| :--- | :--- | :--- |
| A | 20 坪 | 500 萬 |
| B | 30 坪 | 750 萬 |
| C | 40 坪 | 1,000 萬 |
| D | 50 坪 | 1,250 萬 |

你有沒有發現？

坪數越大，房價似乎越高。

人類可以看出這個規律，機器學習也可以透過資料學習這種關係。

假設今天出現一間 60 坪的房子，電腦就可以根據之前學習到的規律，預測它的房價。

這就是機器學習的基本概念。

#### 3. 你一定要認識的五個專有名詞
##### 名詞 1

Dataset（資料集）

提供給電腦學習的資料。

例如：100 間房子的坪數、屋齡、地點及房價。


##### 名詞 2
Feature（特徵）

用來幫助電腦判斷的資訊。

例如：坪數、屋齡、房間數、地點。

白話：我們提供給電腦參考的條件。


##### 名詞 3
Label（標籤）

我們希望電腦學習預測的正確答案。

例如：房子的實際售價。

白話：我們希望電腦猜出來的答案。


##### 名詞 4
Model（模型）

電腦從資料中學習後，建立出來的預測工具。

白話：就像學生讀完書後，腦袋裡學會的知識。


##### 名詞 5
Prediction（預測）

使用訓練好的模型，對新資料做出判斷。

例如：輸入 60 坪，模型預測房價 1,500 萬。

#### 4. 把整個過程串起來
> **① 準備資料 Dataset**  
> 收集房屋坪數、屋齡、房價

　　↓

> **② 找出特徵 Feature**  
> 坪數、屋齡、地點

　　↓

> **③ 訓練模型 Training**  
> 讓電腦從資料中學習規律

　　↓

> **④ 建立模型 Model**  
> 學會坪數與房價之間的關係

　　↓

> **⑤ 預測 Prediction**  
> 輸入新房子的資料，預測房價


#### 5. 第一次看機器學習程式碼
我們先不寫複雜程式。

假設你使用 Google Colab（可以直接在瀏覽器寫 Python 的工具）。

先看這段：

```python
# 房屋坪數 (建立一個叫做 size 的變數，並把 30 放進去。)
size = 30

# 每坪價格 (建立一個盒子，名字叫做 price_per_ping，裡面放 25。)
price_per_ping = 25

# 計算房屋總價 (* 是乘法)
total_price = size * price_per_ping

# print() 的意思是把結果顯示出來
print(total_price)
```
